```python
import pandas as pd
from pathlib import Path

import mlflow
```

```python
OUTPUT_DIR = Path("../outputs")
MLFLOW_DIR = Path("../mlruns").resolve()

mlflow.set_tracking_uri(
    f"file:///{MLFLOW_DIR.as_posix()}"
)

mlflow.set_experiment("AI Text Detection")

print("Tracking URI:", mlflow.get_tracking_uri())
```

```python
ROBERTA_RESULTS_FILE = OUTPUT_DIR / "roberta_results.csv"

roberta_results_df = pd.read_csv(ROBERTA_RESULTS_FILE)

roberta_results_df
```

```python
row = roberta_results_df.iloc[0]

accuracy = float(row["accuracy"])
precision = float(row["precision"])
recall = float(row["recall"])
f1 = float(row["f1_score"])
roc_auc = float(row["roc_auc"])

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1:", f1)
print("ROC-AUC:", roc_auc)
```

```python
with mlflow.start_run(run_name="RoBERTa"):

    # Model / training parameters
    mlflow.log_param("model", "roberta-base")
    mlflow.log_param("max_length", 256)
    mlflow.log_param("learning_rate", 2e-5)
    mlflow.log_param("train_batch_size", 16)
    mlflow.log_param("eval_batch_size", 16)
    mlflow.log_param("epochs", 2)
    mlflow.log_param("weight_decay", 0.01)
    mlflow.log_param("fp16", True)

    # Test metrics
    mlflow.log_metric("test_accuracy", accuracy)
    mlflow.log_metric("test_precision", precision)
    mlflow.log_metric("test_recall", recall)
    mlflow.log_metric("test_f1", f1)
    mlflow.log_metric("test_roc_auc", roc_auc)

    # Save the CSV as an artifact
    mlflow.log_artifact(str(ROBERTA_RESULTS_FILE))

print("RoBERTa results logged to MLflow.")
```

```python
runs = mlflow.search_runs(
    experiment_names=["AI Text Detection"]
)

runs[[
    "tags.mlflow.runName",
    "metrics.test_accuracy",
    "metrics.test_precision",
    "metrics.test_recall",
    "metrics.test_f1",
    "metrics.test_roc_auc"
]]
```
