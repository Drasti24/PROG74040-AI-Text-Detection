```python
import pandas as pd
from pathlib import Path
import mlflow
```

```python
MLFLOW_DIR = Path("../mlruns").resolve()

mlflow.set_tracking_uri(
    f"file:///{MLFLOW_DIR.as_posix()}"
)

mlflow.set_experiment("AI Text Detection")

print("Tracking URI:", mlflow.get_tracking_uri())
```

```python
runs = mlflow.search_runs(
    experiment_names=["AI Text Detection"]
)

runs[[
    "tags.mlflow.runName",
    "metrics.test_accuracy",
    "metrics.test_precision",
    "metrics.test_recall",
    "metrics.test_f1",
    "metrics.test_roc_auc"
]]
```

```python
with mlflow.start_run(run_name="DeBERTa"):

    # Parameters
    mlflow.log_param("model", "microsoft/deberta-base")
    mlflow.log_param("max_length", 256)
    mlflow.log_param("learning_rate", 2e-5)
    mlflow.log_param("train_batch_size", 16)
    mlflow.log_param("epochs", 1)
    mlflow.log_param("training_samples", 100000)
    mlflow.log_param("validation_samples", 20000)
    mlflow.log_param("fp16", False)

    # Original test metrics
    mlflow.log_metric("test_accuracy", 0.996087)
    mlflow.log_metric("test_precision", 0.991243)
    mlflow.log_metric("test_recall", 0.998310)
    mlflow.log_metric("test_f1", 0.994764)
    mlflow.log_metric("test_roc_auc", 0.999901)

    # HC3 external metrics
    mlflow.log_metric("hc3_accuracy", 0.498594)
    mlflow.log_metric("hc3_precision", 0.397516)
    mlflow.log_metric("hc3_recall", 0.999996)
    mlflow.log_metric("hc3_f1", 0.568831)
    mlflow.log_metric("hc3_roc_auc", 0.979550)

print("DeBERTa logged.")
```

```python
with mlflow.start_run(
    run_name="TF-IDF Logistic Regression - Final"
):

    mlflow.log_param(
        "model",
        "TF-IDF + Logistic Regression"
    )

    # Original test
    mlflow.log_metric("test_accuracy", 0.994992)
    mlflow.log_metric("test_precision", 0.993312)
    mlflow.log_metric("test_recall", 0.993239)
    mlflow.log_metric("test_f1", 0.993275)
    mlflow.log_metric("test_roc_auc", 0.999743)

    # HC3
    mlflow.log_metric("hc3_accuracy", 0.469753)
    mlflow.log_metric("hc3_precision", 0.378184)
    mlflow.log_metric("hc3_recall", 0.935683)
    mlflow.log_metric("hc3_f1", 0.538655)
    mlflow.log_metric("hc3_roc_auc", 0.758304)

print("Logistic final run logged.")
```

```python
with mlflow.start_run(
    run_name="TF-IDF SVM - Final"
):

    mlflow.log_param(
        "model",
        "TF-IDF + LinearSVC"
    )

    # Original test
    mlflow.log_metric("test_accuracy", 0.999371)
    mlflow.log_metric("test_precision", 0.999192)
    mlflow.log_metric("test_recall", 0.999118)
    mlflow.log_metric("test_f1", 0.999155)
    mlflow.log_metric("test_roc_auc", 0.999994)

    # HC3
    mlflow.log_metric("hc3_accuracy", 0.561785)
    mlflow.log_metric("hc3_precision", 0.422031)
    mlflow.log_metric("hc3_recall", 0.878491)
    mlflow.log_metric("hc3_f1", 0.570157)
    mlflow.log_metric("hc3_roc_auc", 0.764126)

print("SVM final run logged.")
```
